```python
import pandas as pd
import numpy as np
import networkx as nx
import toml
```

```python
demand_path = "../case_studies/stylized_EU/inputs/demand.csv"
generation_availability_path = "../case_studies/stylized_EU/inputs/generation_availability.csv"
generation_path = "../case_studies/stylized_EU/inputs/generation.csv"
scalars_path = "../case_studies/stylized_EU/inputs/scalars.toml" 
```

```python
demand_df = pd.read_csv(demand_path)
generation_availability_df = pd.read_csv(generation_availability_path)  
generation_df = pd.read_csv(generation_path)

with open(scalars_path, 'r') as f:
    scalars = toml.load(f)

value_of_lost_load = scalars['value_of_lost_load']
relaxation = scalars['relaxation']
```

```python
print(f"Demand {demand_df.shape} Generation Avail {generation_availability_df.shape} Generation {generation_df.shape}")
print(f"Value of lost load: {value_of_lost_load} Relaxation: {relaxation}")
```

```python

tech_per_location = (
    generation_availability_df
        .groupby('location')['technology']
        .nunique()
        .sort_values(ascending=False)   # optional, just for tidy output
)

# How many different technologies overall (across all locations)?
total_unique_tech = generation_availability_df['technology'].nunique()

print("Technologies per location:\n", tech_per_location, "\n")
print("Total unique technologies across every location:", total_unique_tech)
```

```python

tech_lists = (generation_availability_df
              .groupby('location')['technology']
              .unique())        # each element is a NumPy array of tech names
print("Tech lists:\n", tech_lists, "\n")
```
